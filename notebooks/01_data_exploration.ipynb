## Data Quality Assessment

Before cleaning the dataset, we examine missing values, cancellations, duplicate records, and unusual quantity and price values. These findings will be used to define explicit cleaning rules for the forecasting pipeline.

```python
total_rows = len(retail_2009_2010)

cancelled_mask = (
    retail_2009_2010["Invoice"]
    .astype(str)
    .str.startswith("C")
)

print("DATA QUALITY SUMMARY")
print("-" * 40)

print(f"Total rows: {total_rows:,}")

print("\nMissing values:")
print(retail_2009_2010.isna().sum())

print("\nCancellations:")
print(f"Cancelled rows: {cancelled_mask.sum():,}")
print(f"Cancelled row percentage: {cancelled_mask.mean() * 100:.2f}%")
print(
    f"Unique cancelled invoices: "
    f"{retail_2009_2010.loc[cancelled_mask, 'Invoice'].nunique():,}"
)

print("\nQuantity:")
print(f"Minimum quantity: {retail_2009_2010['Quantity'].min():,}")
print(f"Maximum quantity: {retail_2009_2010['Quantity'].max():,}")
print(
    f"Rows with quantity <= 0: "
    f"{(retail_2009_2010['Quantity'] <= 0).sum():,}"
)

print("\nPrice:")
print(f"Minimum price: {retail_2009_2010['Price'].min()}")
print(f"Maximum price: {retail_2009_2010['Price'].max()}")
print(
    f"Rows with price <= 0: "
    f"{(retail_2009_2010['Price'] <= 0).sum():,}"
)

print("\nDuplicates:")
print(f"Duplicate rows: {retail_2009_2010.duplicated().sum():,}")

print("\nDataset range:")
print(f"First transaction: {retail_2009_2010['InvoiceDate'].min()}")
print(f"Last transaction: {retail_2009_2010['InvoiceDate'].max()}")
```

## Non-Product Transaction Investigation

Some transaction rows represent administrative charges, fees, postage, or manualadjustments rather than merchandise sales. These records must be identified so they do not contribute to product-demand forecasts.

```python
special_terms = (
    "POSTAGE|CARRIAGE|BANK CHARGES|AMAZON"
    "MANUAL|ADJUST|DISCOUNT|DOTCOM"
)

special_rows = retail_2009_2010[
    retail_2009_2010["Description"]
    .str.contains(
        special_terms,
        case=False,
        regex=True,
        na=False
    )
]

print("Potential non-product rows:", len(special_rows))

special_rows[
    ["StockCode", "Description"]
].drop_duplicates().sort_values(
    ["StockCode", "Description"]
).head(50)
```

## Administrative Stock Code Assessment

Keyword matching on product descriptions produced false positives because legitimate merchandise can contain terms such as "carriage" or "dotcom." Administrative transactions are therefore identified using explicit StockCodes rather than broad description matching.

```python
administrative_codes = {
    "ADJUST",
    "ADJUST2",
    "B",
    "BANK CHARGES",
    "C2",
    "D",
    "DOT",
    "M",
    "POST",
    "AMAZONFEE",
}

stock_codes = retail_2009_2010["StockCode"].astype(str)

administrative_mask = (
    stock_codes.isin(administrative_codes)
    | stock_codes.str.startswith("gift_")
)

print("Administrative/non-merchandise rows:", administrative_mask.sum())

retail_2009_2010.loc[
    administrative_mask,
    ["StockCode", "Description"]
].drop_duplicates().sort_values(
    ["StockCode", "Description"]
)
```
