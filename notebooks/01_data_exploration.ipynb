# Online Retail II - Exploratory Data Analysis

This notebook perfoms an initial inspection of the Online Retail II dataset to understand its structure, data types, 
missing values, cancellations, duplicates, and unusual transaction values before
building the demand forecasting pipeline.

```python
import pandas as pd
from pathlib import Path
```

```python
# Cell 2
DATA_PATH = Path("../data/raw/online_retail_II.xlsx")

print(DATA_PATH.exists())
```

```python
excel_file = pd.ExcelFile(DATA_PATH)

excel_file.sheet_names
```

```python
retail_2009_2010 = pd.read_excel(
    DATA_PATH,
    sheet_name="Year 2009-2010"
)
```
